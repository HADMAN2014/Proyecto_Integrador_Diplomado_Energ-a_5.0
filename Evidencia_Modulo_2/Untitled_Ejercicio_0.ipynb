```python
import pandas as pd

file_path = '/content/Sistematización_General_Retroalimentaciones_AAP__CIDEMOS_2026_-_all_versions_-_labels_-_2026-09-30-17-01-52.xlsx'
df = pd.read_excel(file_path)

print('First 5 rows of the DataFrame:')
display(df.head())

print('\nDataFrame Info:')
df.info()
```

```python
# Convert '1. Fecha:' to datetime objects, coercing errors to NaT
df['1. Fecha:'] = pd.to_datetime(df['1. Fecha:'], errors='coerce')

# Convert '8. Edad del o la participante:' to numeric, coercing errors to NaT
df['8. Edad del o la participante:'] = pd.to_numeric(df['8. Edad del o la participante:'], errors='coerce')

# Filter for 'Departamento del Cauca'
df_filtered = df[df['3. Departamento:'] == 'Cauca'].copy()

# Filter for 'Municipio de Buenos Aires y Toribio'
df_filtered = df_filtered[df_filtered['4. Municipio de Cauca'].str.strip().isin(['Buenos Aires', 'Toribio'])].copy()

# Filter for 'Proyecto Acuerdo de Cuidado y Proteccion'
df_filtered = df_filtered[df_filtered['2. Proyecto:'].str.strip() == 'Acuerdos de cuidado y protección'].copy()

# Drop rows where '8. Edad del o la participante:' is NaN after conversion for age filtering
df_filtered_age = df_filtered.dropna(subset=['8. Edad del o la participante:']).copy()

# Filter ages 10-14
df_filtered_10_14 = df_filtered_age[(df_filtered_age['8. Edad del o la participante:'] >= 10) & (df_filtered_age['8. Edad del o la participante:'] <= 14)].copy()

# Filter ages 15-18
df_filtered_15_18 = df_filtered_age[(df_filtered_age['8. Edad del o la participante:'] >= 15) & (df_filtered_age['8. Edad del o la participante:'] <= 18)].copy()

# Filter for September dates
df_filtered_september = df_filtered[df_filtered['1. Fecha:'].dt.month == 9].copy()

print("DataFrame filtered by Departamento, Municipio, and Proyecto:")
display(df_filtered.head())
print(f"Number of rows after filtering: {len(df_filtered)}")

print("\nDataFrame filtered by ages 10-14:")
display(df_filtered_10_14.head())
print(f"Number of rows after filtering (10-14): {len(df_filtered_10_14)}")

print("\nDataFrame filtered by ages 15-18:")
display(df_filtered_15_18.head())
print(f"Number of rows after filtering (15-18): {len(df_filtered_15_18)}")

print("\nDataFrame filtered by September dates:")
display(df_filtered_september.head())
print(f"Number of rows after filtering (September): {len(df_filtered_september)}")
```

```python
print("Unique values in '8. Edad del o la participante:' after initial filtering:")
print(df_filtered['8. Edad del o la participante:'].unique())
print("\nData type of '8. Edad del o la participante:' after initial filtering:")
print(df_filtered['8. Edad del o la participante:'].dtype)
```
